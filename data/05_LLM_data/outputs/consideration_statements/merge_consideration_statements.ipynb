```python
import pandas as pd
import os
```

```python
# map each political‐leaning label to its CSV path
file_map = {
    "Left":            "left_consideration_statements.csv",
    "Left-Liberal":    "leftliberal_consideration_statements.csv",
    "Centrist":        "centrist_consideration_statements.csv",
    "Right-Liberal":   "rightliberal_consideration_statements.csv",
    "Right":           "right_consideration_statements.csv",
}

frames = []
for leaning, path in file_map.items():
    # read
    df = pd.read_csv(path)

    # ensure column exists
    col = "Statements, similarity_score_5" # Define Column Name for extraction - Here: Health Prevention
    if col not in df.columns:
        raise KeyError(f"Missing column {col!r} in {path}")

    # select + drop truly empty rows
    sub = df[[col]].copy()
    # drop NaN
    sub = sub.dropna(subset=[col])
    # drop empty or whitespace-only
    sub = sub[sub[col].astype(str).str.strip() != ""]

    # rename & tag
    sub = sub.rename(columns={col: "Health Prevention Consideration Statements"})
    sub["Political Leaning"] = leaning

    # reset index for a clean concat
    frames.append(sub.reset_index(drop=True))

# concatenate all leanings
merged_df = pd.concat(frames, ignore_index=True)

# sanity-check
print(merged_df["Political Leaning"].value_counts())
print("Any missing statements?", merged_df["Health Prevention Consideration Statements"].isna().sum())

# view result
print(merged_df.head())
```
